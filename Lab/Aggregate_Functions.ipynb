```python
import pandas as pd
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Value': [10, 15, 20, 25, 30, 35],
    'Fee' :[20000,25000,26000,22000,24000,35000],
    'Duration':['30day','40days','35days','40days','60days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
total_sum = df['Value'].aggregate('sum')
print("Total Sum:", total_sum)
```

```python
df[['Value']].cumsum()
```

```python
total_mean = df['Value'].aggregate('mean')
print("Total Mean:", total_mean)
```

```python
max_value = df['Value'].aggregate('max')
print("Maximum Value:", max_value)
```

```python
count = df['Value'].aggregate('count')
print("Count of Values:", count)
```

```python
std = df['Value'].aggregate('std')
print("Standard Deviation:", std)
```

```python
result = df[['Fee', 'Discount']].aggregate('sum')
print(result)   
```

```python
result2 = df[['Fee', 'Discount']].aggregate('sum')  
print(result2)
```

```python
result_group = df.groupby('Category')[['Fee', 'Discount']].aggregate('sum')
print(result_group)
```
